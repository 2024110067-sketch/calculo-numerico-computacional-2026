```python
def newton_method(f, df, x0, tol=1e-6, max_iter=100):
    """
    Implementa o método de Newton-Raphson para encontrar a raiz de f(x).
    """
    x = x0
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)

        # Evita divisão por zero
        if dfx == 0:
            print("Derivada zero. Método falhou.")
            return None

        # Fórmula de Newton-Raphson
        x_new = x - fx / dfx

        # Verifica convergência
        if abs(x_new - x) < tol:
            print(f"Convergiu em {i+1} iterações.")
            return x_new

        x = x_new

    print("Número máximo de iterações atingido.")
    return x

# Definição da função e sua derivada
f = lambda x: x**3 - x - 2
df = lambda x: 3*x**2 - 1

# Palpite inicial
x0 = 1.5

# Execução
raiz = newton_method(f, df, x0)
print(f"Raiz aproximada: {raiz}")
print(f"f(raiz): {f(raiz)}")
```
